```python
from pathlib import Path
from collections import Counter
import pandas as pd
```

# Data

```python
DATASET_DIR = Path("dataset/")

def get_partition_dataset(which: str) -> tuple[pd.DataFrame, pd.DataFrame]:
    partition_train = DATASET_DIR / f"datasense_{which}_train_1sec.parquet"
    partition_test = DATASET_DIR / f"datasense_{which}_test_1sec.parquet"
    return (
        pd.read_parquet(partition_train),
        pd.read_parquet(partition_test),
    )
```

# Full data

```python
train, test = get_partition_dataset("full")
```

```python
train.info(verbose=True)
```

```python
train_attack_count = Counter(train["label_extended"])
test_attack_count = Counter(test["label_extended"])
padding = max(len(atk) for atk in train_attack_count)
for attack, count in train_attack_count.most_common():
    train_perc = 100 * count / train.shape[0]
    test_perc = 100 * test_attack_count[attack] / test.shape[0]
    print(f"{attack.rjust(padding)} : train {train_perc:05.2f}% | test {test_perc:05.2f}%")
```
